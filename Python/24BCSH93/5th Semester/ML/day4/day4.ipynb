# KNN Algorithm Implementation

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import math
from collections import Counter

data = pd.read_csv("Data.csv")
data
```

```python
# Step 2: Define Euclidean distance function
def euclidean_distance(p1, p2):
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(p1, p2)))
```

```python
#Step 3: Define KNN function
def knn_classify(train_data, y_train, test_point, k):
    distances = []
    # for row in train_data:
    #     features = [row["X1"], row["X2"], row["X3"]]
    #     dist = euclidean_distance(features, test_point)
    #     distances.append((dist, row["Y"]))

    for i in range(len(train_data)):
        dist = euclidean_distance(test_point, train_data[i])
        distances.append((dist, y_train[i]))

    print(distances)
    # Step 4: Sort distances and pick k nearest
    distances.sort(key=lambda x : x[0])
    k_nearest = [label for _, label in distances[:k]]

    # Step 5: Majority vote
    most_common = Counter(k_nearest).most_common(1)
    return most_common[0][0]
```

```python
train_data = data.iloc[:, 1:-1].values
y_train = data.iloc[:, -1].values

# Step 6: Example test point
test_point = [0, 0, 0]

print(train_data[0])
print(test_point)
k = 3
predicted_label = knn_classify(train_data, y_train, test_point, k)
```

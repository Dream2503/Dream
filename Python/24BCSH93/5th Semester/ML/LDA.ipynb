```python
import numpy as np
```

```python
c1 = np.array([(4, 1), (2, 4), (2, 3), (3, 6), (4, 4)])
c2 = np.array([(9, 10), (6, 8), (9, 5), (8, 7), (10, 8)])
c1, c2
```

class 1

```python
mu1 = c1.mean(axis=0)
mu2 = c2.mean(axis=0)
mu1, mu2
```

```python
s1 = (c1 - mu1).T @ (c1 - mu1) / c1.shape[0]
s2 = (c2 - mu2).T @ (c2 - mu2) / c2.shape[0]
s1, s2
```

```python
sw = s1 + s2
sw
```

```python
sb = np.outer(mu1 - mu2, mu1 - mu2)
sb
```

```python
mat = np.linalg.inv(sw) @ sb
mat
```

```python
lambda_ = np.max(np.real(np.linalg.eigvals(mat)))
lambda_
```

```python
mat1 = mat - np.eye(2) * lambda_
mat1
```

```python
v = np.linalg.solve(mat1, np.zeros(mat1.shape[0]))
v
```

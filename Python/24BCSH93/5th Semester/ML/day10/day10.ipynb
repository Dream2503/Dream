# Naive Bayes Classification

```python
# importing libraries and the dataset
import pandas as pd

df = pd.read_csv('play_tennis.csv')
df
```

```python
# counting Yes and No values
class_counts = df['play'].value_counts()
class_counts
```

```python
# counting attribute values for each class
for column in df.columns[:-1]:
    print("\n", column)
    print(pd.crosstab(df[column], df['play']))
```

```python
# new instance
X = {'outlook': 'Sunny', 'temp': 'Cool', 'humidity': 'High', 'wind': 'Strong'}

# conditional probabilities
for cls in class_counts.index:
    print("\nProbabilities for play =", cls)

    for attribute, value in X.items():
        count = len(df[(df[attribute] == value) & (df['play'] == cls)])
        total = class_counts[cls]
        probability = count / total
        print(attribute, "=", value, ":", count, "/", total, "=", round(probability, 4))
```

```python
# calculating Naive Bayes probabilities
probabilities = {}

for cls in class_counts.index:
    # prior probability
    probability = class_counts[cls] / len(df)

    # multiply by conditional probabilities
    for attribute, value in X.items():
        count = len(df[(df[attribute] == value) & (df['play'] == cls)])
        total = class_counts[cls]
        probability *= count / total

    probabilities[cls] = probability
    print("Probability for", cls, "=", probability)
```

```python
# final prediction
prediction = max(probabilities, key=probabilities.get)
print("Final Prediction:", prediction)
```

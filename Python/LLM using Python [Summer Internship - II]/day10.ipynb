```python
import matplotlib.pyplot as plt
import numpy as np
import torchvision.transforms as transforms
import torchvision.utils as vutils
from torch.utils.data import DataLoader
from torchvision.datasets import ImageFolder
```

```python
# Define transformation for the dataset
transform = transforms.Compose(
    [
        transforms.Resize(64),  # Resize images to 64x64
        transforms.CenterCrop(64),  # Crop the central part of the image
        transforms.ToTensor(),  # Convert images to tensors
        transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])  # Normalize the images
    ]
)
```

```python
import kagglehub

path = kagglehub.dataset_download("jessicali9530/lfw-dataset", output_dir="./data", force_download=True)
dataset = ImageFolder(root="./data/lfw-deepfunneled/lfw-deepfunneled", transform=transform)
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)
```

```python
# Function to unnormalize and display images
def imshow(img):
    img = img / 2 + 0.5  # unnormalize
    npimg = img.numpy()
    plt.imshow(np.transpose(npimg, (1, 2, 0)))
    plt.show()
```

```python
# Get a batch of images
dataiter = iter(dataloader)
images, _ = next(dataiter)
```

```python
# Show images
imshow(vutils.make_grid(images))
```
